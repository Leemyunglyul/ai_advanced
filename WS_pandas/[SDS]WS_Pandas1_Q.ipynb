### Series 값 수정 및 추가

```python
# 그대로 실행하세요.
s1 = pd.Series([1,2,3,4,5], index=list('ABCDE'))
s2 = pd.Series([1,2,3,4,5])
display(s1)
display(s2)
```

```python
# s1의 c 인덱스의 값(3)을 30으로 수정하세요.
s1['C'] = 30
s1
```

```python
# s2의 세 번째 값(3)을 10으로 나눠서 저장하세요.
s2[2] = s2[2] / 10
s2
```

```python
# s1과 s2를 더해서 s3에 저장하세요.
s3 = s1 + s2
s3
```

```python
# s2의 index를 A B C D E로 변경한 후 다시 더해서 s4에 저장하세요.
s2.index = list('ABCDE')
s4 = s1 + s2
s4
```

```python
# s1 전체에 10을 더하세요.
s1 = s1+10
s1
```

```python
# s2 전체에 10을 곱하세요.
s2 = s2 * 10
```

```python
s2
```

```python
# s1이 s2보다 큰지 비교하세요.
s1 > s2
```

```python
# s2의 값이 10보다 작은지 비교하세요.
s2 < 10
```

```python
# s1과 s2의 차를 출력하고 양수인지 비교하세요.
display(s1-s2)
display((s1-s2)>0)
```

```python
# s2가 짝수인지 비교하세요.
s2 % 2 ==0
```

```python
# s1이 홀수이면서 3의 배수인지 비교하세요.
(s1 % 2 == 1) & (s1 % 3 == 0)
```

```python
# s1의 index의 값이 'A'인지 비교하세요.
s1.index == 'A'
```

### DataFrame 생성 및 값 조회

#### 예시
image.png

```python
# 위 예시와 동일한 DataFrame을 리스트를 이용하여 생성하고 df에 저장하세요.
data = [
    ['월', '돈까스', 9000],
    ['화', '비빕밥', 7500],
    ['수', '냉면', 8000],
    ['목', '피자', 9900],
    ['금', '짜장면', 7000]
]

df = pd.DataFrame(data, columns=['요일', '메뉴', '가격'])
df
```

```python
# 위 예시와 동일한 DataFrame을 Dictionary를 이용하여 생성하고 df에 저장하세요.
data = {
    '요일': ['월', '화', '수', '목', '금'],
    '메뉴': ['돈까스', '비빕밤', '냉면', '피자', '짜장면'],
    '가격': [9000, 7500, 8000, 9900, 7000],
}
df = pd.DataFrame(data)
df
```

```python
# df의 열 이름을 요일->Dayofweek, 메뉴->Menu, 가격->Price으로 변경하세요.
df.columns = ['Dayofweek', 'Menu', 'Price']
df
```

```python
# df의 열 이름 중 Dayofweek를 Dow로 변경하세요.
df.rename(columns={'Dayofweek': 'Dow'}, inplace=True)
df
```

```python
# df의 Dow 열을 추출하세요. (출력 형태 시리즈)
df['Dow']
```

```python
# df의 마지막 열을 추출하세요.
df['Price']
```

```python
# Dow 열을 추출하세요. (출력 형태 데이터프레임)
df[['Dow']]
```

```python
# 1~2번 행을 추출하세요.
df[1:3]
```

```python
# 월요일의 점심 메뉴를 출력하세요.
df['Menu'][0]
```

```python
# 사실 월요일 점심 메뉴는 돈까스가 아니었습니다.
# 월요일의 점셈 메뉴를 물로 변경하세요.
df['Menu'][0] = '물'
df
```

```python
# 월요일의 음식 가격에서 7000을 빼세요
df['Price'][0] -=7000
df
```

```python
# 새 열 '비고'를 추가하고 값을 모두 '점심'으로 설정하세요.
df['비고'] = '점심'
df
```

#### DataFrame 연산

```python
# df의 Price 열은 원 단위입니다. Price열의 값을 1000으로 나눈 값으로 천원 단위 열인 Price(천원)을 생성하세요.
df['Price(천원)'] = df['Price']/1000
df
```

```python
# df의 2번 행을 선택하세요.
# hint. 전치 + 열 선택
df.T[2]
```

```python
# df의 Dow~Price 열을 선택하세요.
# hint. 팬시 혹은 전치+슬라이싱
df.T['Menu':'Price'].T
```

```python
# df에 새 열 "인원"을 추가합니다.
# 인원은 위에서부터 순서대로 [1, 2, 2, 4, 3] 명을 입력하시면 됩니다.
df['인원'] = [1, 2, 2, 4, 3]
df
```

```python
# df에 새 열 "총합"을 추가합니다.
# 총합은 Price * 인원입니다.
df['총합'] = df['Price'] * df['인원']
df
```

```python
# df에 새 열 "총합(만원)"을 추가합니다.
df['총합(만원)'] = df['총합'] / 10000
df
```

```python
# df의 Price 값이 7000원 이상이면서 총합이 20000원 이상인지 확인하세요.
(df['Price'] >= 7000) & (df['총합'] >= 20000)
```
