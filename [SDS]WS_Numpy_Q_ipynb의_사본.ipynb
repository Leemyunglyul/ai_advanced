### 배열 형태 변경

```python
arr = np.array([[1,4,5], [5,6,7]])
arr
```

 - arr를 1D 배열로 변경 후 arr1d에 저장 후 배열 및 shape 출력

```python
arr1d = arr.flatten()
print(arr1d)
print(arr1d.shape)
```

```python
arr1d = arr.reshape(-1)
print(arr1d)
print(arr1d.shape)
```

```python
arr1d = arr.ravel()
print(arr1d)
print(arr1d.shape)
```

- arr1d를 3행 2열의 2D 배열로 변경 후 arr2d에 저장 후 배열 및 shape 출력

```python
arr2d = arr1d.reshape(3,2)
print(arr2d)
print(arr2d.shape)
```

 - arr1d를 열이 1개인 2D 배열로 변경 후 arr2d에 저장 후 배열 및 shape 출력

```python
arr2d = arr1d.reshape(-1, 1 )
print(arr2d)
print(arr2d.shape)
```

### 기술 통계 및 기타 함수

```python
np.random.seed(1024)
idx = np.random.choice(25, 5, replace=False)
arr = np.random.randint(1, 100, 25)
arr[idx] = 0
arr = arr.reshape(5, -1)
arr
```

- arr의 전체 합을 출력하시오

```python
arr.sum()
```

- arr의 열 별 합을 출력하시오

```python
arr.sum(axis=0)
```

 - arr의 열 별 평균을 출력하시오

```python
arr.mean(axis=0)
```

 - arr의 열 별 모분산을 출력하시오

```python
arr.var(axis=0)
```

 - arr의 열 별 표본분산을 출력하시오

```python
arr.std(axis=0)
```

 - arr의 열 별 표분표준편차를 소수점 3자리까지 출력하시오

```python
arr.std(axis=0).round(3)
```

 - arr의 열 별 표분표준편차를 출력하되, 분산 함수와 제곱근 함수를 이용하시오
   - np.sqrt(arr)

```python
np.sqrt(arr.var(axis=0))
```

- arr의 열 별 합의 합을 출력하시오

```python
arr.sum(axis=0).sum()
```

- arr의 행 별 중앙값을 출력하시오

```python
np.median(arr, axis=1)
```

- arr의 행 별로 가장 큰 값을 출력하시오

```python
arr.max(axis=1)
```

 - arr의 행 별 가장 큰 값의 위치(index)를 idx 변수에 저장하고 출력하시오

```python
idx = arr.argmax(axis=1)
idx
```

- arr의 전체에 한 개라도 참인 값이 있는 지 확인하시오

```python
arr.any()
```

- arr의 열 중 모든 값이 참인 열이 있는 지 확인하시오

```python
arr.all(axis=0)
```

 - arr의 열 중 거짓 값이 하나라도 포함된 열이 있는 지 확인하시오

```python
~ arr.all(axis = 0)
```
