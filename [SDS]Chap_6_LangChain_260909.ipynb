## OpenAI API 키 설정

```python
import os

os.environ["OPENAI_API_KEY"] = ''
```

## Prompt 생성

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template("다음 주제에 대해 설명해 주세요 : {subject}")
prompt
```

## Model 생성

```python
from langchain_openai import ChatOpenAI

# ChatOpenAI 인스턴스를 생성하면서 모델과 파라미터를 설정.
llm = ChatOpenAI(
    model_name="gpt-5.4-mini",  # 사용할 모델을 "gpt-5.4-mini"로 지정.
    max_tokens=2048,           # 모델이 반환할 최대 토큰 수
)
```

## Output Parser 생성

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
parser
```

## 체인 구성 및 실행

```python
chain = prompt | llm | parser

# 인보크
chain.invoke({'subject':'인공지능'})
```

## 스트림 생성 및 출력

```python
# '인공지능' 토픽에 대한 스트림을 생성하고 반복합니다.
for token in chain.stream({'subject':'인공지능'}):
    print(token, end='', flush=True)
```

## 멀티모달 추론 예시

```python
from gmteacher import download_file

# 이미지 다운로드
download_file('IMAGE')
```

```python
import base64

# base64 모듈을 이용하여 인코딩
def encode_data(data_path):
    with open(data_path, "rb") as data_file:
        return base64.b64encode(data_file.read()).decode("utf-8")

encoded_image = encode_data('./data/seoul_night.jpg')
```

```python
# 프롬프트 템플릿 생성
from langchain_core.prompts import ChatPromptTemplate

# 이미지 분석을 위한 시스템 프롬프트
system_prompt = '''
당신은 유능한 이미지 분석가 AI입니다.
전달된 이미지 속에 포함된 요소들을 면밀히 관찰하고,
각 요소가 어떤 의미를 갖고 있는지 분석하세요.
이미지의 색상, 구성, 인물의 표정과 자세, 배경 요소, 조명 등
모든 시각적 요소를 분석하여 이미지가 전달하고자 하는 메시지나 감정을 설명하세요.
또한, 이 이미지가 제작된 문화적, 사회적, 역사적 배경을 고려하여
추가적인 해석을 한국어로 제공하세요.
'''
# 프롬프트 템플릿 생성
chat_template = ChatPromptTemplate.from_messages(
    [("system", system_prompt),
     ("human", [{"type":"image_url", "image_url": {"url": "{image_data}"}}])])
```

```python
# 프롬프트에 전달할 변환된 이미지 데이터
image_content = f"data:image/jpg;base64,{encoded_image}"

# 체인 구성 및 실행
chain = chat_template | llm
for token in chain.stream({'image_data': image_content}):
    print(token.content, end='', flush=True)
```

## Chain-of-Thought 생성

```python
email_conversation = """From:  김길동 (gildong.kim@pgpt.com)
To: 이동윤 (dongyoon.lee@pgpt.com)
Subject:  Private GPT 구축 방안 관련 미팅 일정 제안

안녕하세요, 이동윤 프로님,

Private GPT 구축에 필요한 상세한 요구 성능 기술서를 요청드립니다.

특히 구축에 필요한 언어 모델의 사양과 API 혹은 온프레미스 구축 시 예상되는 비용을 상세히 기술해주시기 바랍니다.

해당 자료를 금주 금요일(6월 27일)까지 메일로 발송해주셨으면 좋겠습니다.

또한 가능하시면 해당 자료에 대한 미팅을 차주 화요일(7월 1일) 오전 10시에 진행했으면 좋겠습니다.

감사합니다.

김길동 프로
"""
```

```python
from pydantic import BaseModel, Field
from typing import Dict

# 날짜 타입 스키마
class DateInfo(BaseModel):
    purpose: str = Field(description="날짜의 목적")
    date: str = Field(description="날짜(MM-DD)")

# 구조화된 출력 스키마
class EmailSummary(BaseModel):
    person: str = Field(description="메일을 보낸 사람")
    email: str = Field(description="메일을 보낸 사람의 이메일 주소")
    subject: str = Field(description="메일 제목")
    summary: str = Field(description="메일 본문을 요약한 텍스트")
    date: DateInfo = Field(description="메일 본문에 언급된 주요 날짜")
    think: str = Field(description="추론 과정(step-by-step)")
```

```python
# 프롬프트 템플릿 작성
prompt = PromptTemplate.from_template("""
너는 좋은 AI 도우미야. 지금은 2026년이야.
이메일 내용중 주요 내용을 추출해줘.

# 지시사항:
1. 먼저 메일 본문을 읽고, 언급된 모든 날짜와 그 목적을 논리적으로 추론해줘.
2. 각 날짜가 마감일인지, 미팅인지, 혹은 단순 언급인지 단계를 나누어 생각해줘.
3. 최종 결과는 반드시 제공된 포맷 지침을 따라줘.

# 이메일 내용:
{email_conversation}

# 추론 과정(step-by-step):""")
llm_with_so = llm.with_structured_output(EmailSummary)
chain = prompt | llm_with_so
result = chain.invoke({"email_conversation": email_conversation})
```

```python
print('추론 과정:')
print(result.model_dump()['think'])
```
