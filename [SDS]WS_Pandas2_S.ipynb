```python
# 불리언 인덱싱을 이용하여 2014년 신규 cctv 수가 150개 이상이고
# 동시에 2016년 신규 cctv 수가 300개 이하인 행을 추출하세요.
cond1 = df['2014년'] >=150
cond2 = df['2016년'] <=300

df[cond1&cond2]
```

```python
# loc Indexer를 이용하여 2014년 신규 cctv 수가 150개 이상이고
# 동시에 2016년 신규 cctv 수가 300개 이상인 행을 추출하세요.
cond1 = df['2014년'] >=150
cond2 = df['2016년'] <=300

df.loc[cond1&cond2, :]
```

```python
# iloc Indexer를 이용하여 2014년 신규 cctv 수가 150개 이상이고 
# 동시에 2016년 신규 cctv 수가 300개 이상인 행을 추출하세요.
cond1 = df['2014년'] >=150
cond2 = df['2016년'] <=300

df.iloc[(cond1&cond2).values, :]
```

```python
# df의 데이터 중 2014년을 기준으로 신규 cctv 수가 매년 증가한 지역의 데이터를 출력하세요.
# 2014<2015<2016인 지역의 기관명
cond = (df['2014년'] < df['2015년']) & (df['2015년'] < df['2016년'])
df.loc[cond, :]
```

```python
# df의 데이터 중 2014년을 기준으로 신규 cctv 수가 매년 증가한 지역 외 나머지 지역의 데이터를 출력하세요.
cond = (df['2014년'] < df['2015년']) & (df['2015년'] < df['2016년'])
df.loc[~cond, :]
```

```python
# df의 데이터 중 2014년을 기준으로 신규 cctv 수가 매년 감소한 지역의 기관명을 출력하세요.
# 2014>2015>2016인 지역의 기관명
cond = (df['2014년'] > df['2015년']) & (df['2015년'] > df['2016년'])
df[cond]
```

```python
# 강서구의 2014년 대비 2016년의 신규 cctv의 비율을 백분율로 계산한 후 소수점 셋째자리까지 출력하세요.
강서구2016 = df.loc[df['기관명'] == '강서구', '2016년']
강서구2014 = df.loc[df['기관명'] == '강서구', '2014년']

비율 = (강서구2016/강서구2014*100).iloc[0]
print(f'강서구의 2014년 대비 2016년의 신규 cctv 수 비율은 약 {비율:.3f}%입니다.')
```

```python
# 모든 지역의 2014년 대비 2016년의 신규 cctv 증감율을 계산한 후
# 새 열 "증감률"에 저장하세요.
df['증감률'] = ((df['2016년'] - df['2014년'])/df['2014년']*100)
df
```

```python
# df의 데이터 중 2013년 이전 cctv 수가 2014년~2016년 누적치보다 많은 데이터를 추출하세요.
cond = df['2013년도 이전'] > df['2014년']+df['2015년']+df['2016년']
df[cond]
```

```python
# df의 데이터 중 전체 cctv 누적 cctv 수가 2013년도 이전 cctv수보다 두 배 이상으로
# 증가한 지역의 데이터를 출력하세요.
cond = (df['소계']/df['2013년도 이전']) >=2
df[cond]
```
