## 클래스

````python
class Product:
    title = '자동차' # 클래스 변수
    price = '30000'

    def __init__(self): # 생성자
        pass

    def printInfo(self): # 메소드
        print(f'{self.title}: {self.price}')
p = Product()
p.printInfo()
````

````python
class Product:
    def __init__(self, title, price): # 생성자
        self.title = title # 인스턴스 변수
        self.price = price

    def __del__(self): # 소멸
        print('Instance is deleted')

    def printInfo(self): # 메소드
        print(f'{self.title}: {self.price}')

p = Product('자동차', 20000)
p.printInfo()
p = None
````

````python
class Person:
    def __init__(self, name):
        self.name = name

    def printInfo(self):
        print("name:", self.name)

class Student(Person): # Person 클래스 상속
    def __init__(self, name, hakbun):
        super().__init__(name) # Person(부모) 클래스 인스턴스 초기화
        self.hakbun = hakbun

    def printInfo(self): # 오버라이딩
        super().printInfo()
        print("hakbun:", self.hakbun)

p = Person('김')
p.printInfo()

s = Student('박', '2020')
s.printInfo()
````
