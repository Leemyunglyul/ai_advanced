## 배열 형태 변경

```python
# shape 변경
array = np.arange(16) # 0~15의 정수 1D 배열
array, array.shape
```

```python
# array.reshape(축 별 크기, ...)
# np.reshape(arr, shape)
# 지정한 축 별 크기 또는 shape의 size가 array의 size와 동일
# array.shape 메소드
arr1 = array.reshape(4, 4)
arr1, arr1.shape
```

```python
# np.reshape 함수
np.reshape(array, (4, 4))
```

```python
# 1개 축만 -1 설정 가능 (자동 계산)
array.reshape(-1, 8), array.reshape(2, -1).shape
```

```python
arr1 = array.reshape(4, 4)
arr1
```

```python
arr1.reshape(-1) # 1d 변경 -> view
```

```python
arr1.ravel() # 1d 변경 -> view
```

```python
arr1.flatten() # 1d 변경 -> copy
```

```python
arr1 = array # arr1과 array 동일 (reference)
```

```python
# data는 공유, 속성은 따로
# reshape, ravel이 view 생성
arr1 = array.view()
```

```python
arr2 = array.reshape(4, 4)
arr2[0, 0] = 1000 # view라서 데이터를 공유
arr2, array
```

```python
# flatten은 copy
arr1 = array.copy() # 속성, 데이터 모두 따로
```

### 불리언 인덱싱

```python
# 배열 연산 (element-wise)
# 동일 인덱스 (위치)끼리 연산
arr1 = np.arange(1, 5).reshape(2,2)
arr2 = np.arange(10, 50, 10).reshape(2,2)
arr1 - arr2 # 사칙, 비교 연산 모두 가능
```

```python
arr1 = np.array([[1,2],[3,4]])
arr2 = np.array([10, 20])
arr1.shape, arr2.shape
arr1+arr2
```

```python
# Broadcast: shape이 다른 배열끼리 연산
# (3, 1, 2), (1, 4, 2)
# 같은 축(번호)의 크기가 동일 또는 한쪽 1
arr1 = np.ones((3, 1, 2))
arr2 = np.ones((1, 4, 2))
arr1 + arr2, (arr1+arr2).shape
```

```python
arr1 % 2 == 0 # 배열과 스칼라 간의 연산 가능
```

```python
a = 10

a % 2 == 0 and a > 5 # 파이썬 논리 연산
```

```python
cond1 = arr1 % 2 == 0
cond2 = arr1 > 5
cond1 and cond2 # 파이썬 논리 연산자 사용 불가
```

```python
# 비트 연산자를 논리 대신 사용
cond1 = arr1 % 2 == 0
cond2 = arr1 > 5
cond1 & cond2
```

```python
(arr1 % 2 == 0) & (arr1 > 5)
```

```python
arr1[1] = 1000
arr1
```

```python
array
```

```python
# 인덱스 자리에 bool 타입의 리스트/터플/배열
bidx = array % 2 == 0
bidx
```

```python
array[bidx] # 결과는 1D
```

```python
# 넘파이 불리언 인덱싱은 주로 1D에 사용
array[array > 5] = 1000
array
```
