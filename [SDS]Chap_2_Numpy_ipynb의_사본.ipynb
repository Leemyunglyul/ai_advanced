## 배열 속성

```python
array = np.array([[1, 2], [3, 4]])
print('array의 차원:', array.ndim)
```

```python
print('array의 각 차원별 크기:', array.shape)
```

```python
print('array의 요소의 총 개수 :', array.size)
```

```python
print('array의 데이터 유형 :', array.dtype)
```

```python
print('array의 전치 (전) :')
print(array)
print('array의 전치 (후) :')
print(array.T)
```

## 배열 형태 변경

```python
array.reshape(4, 4), array.reshape(4, 4).shape
```

```python
array.reshape(-1, 8), array.reshape(-1, 8).shape
```

## 배열 데이터 타입 변경

```python
x = np.array([[1,2], [3,4]])
x, x.dtype
```

```python
x = np.array([[1,2], [3,4]], dtype=np.float64)
x, x.dtype
```

```python
x.astype(np.int32)
```

```python
np.int32(x)
```

```python
x.astype('int')
```

## 배열 인덱싱

### 정수 인덱싱

```python
array = np.arange(9).reshape(3, 3)
array[0]
```

```python
array[-1]
```

```python
array[0, 0]
```

```python
array[-1, -1]
```

### 불리언 인덱싱

```python
bidx = array % 2 == 0
bidx
```

```python
array[bidx]
```

```python
array[array > 5]
```

### 팬시 인덱싱

```python
row = [0, 1]
col = [2, 1]
array[row, col]
```

```python
array[[0,1], [2, 1]]
```

```python
array[0, 2], array[1, 1]
```

### 슬라이싱

```python
array[0, :]
```

```python
array[:, 1]
```

```python
array[:2, 1:]
```

```python
array[-1:, 1:3]
```
