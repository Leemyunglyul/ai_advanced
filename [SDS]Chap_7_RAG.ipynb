## PDF 로더 생성

## PDF 파일 로딩 및 문서 수 확인

```python
# PDF 로더
docs = loader.load()

# 로드된 문서의 수 확인
len(docs)
```

## 데이터 전처리

```python
import re
# \x07(벨), \t(탭), \s{2,} 공백 2개 이상을 모두 공백 1개(' ')로 치환
pattern = re.compile(r'[\x07\t]|\s{2,}')

# 모든 문서에 적용
for doc in docs:
    doc.page_content = pattern.sub(' ', doc.page_content).strip()
```

```python
docs[4] # 재확인
```

## 문서 분할 및 문서 확인

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 문자열 분할기 설정
text_splitter = RecursiveCharacterTextSplitter(chunk_size=600, chunk_overlap=0)

# 문서 분할 (1) : 문서 로드 과정에서 텍스트 분할기를 함께 적용
split_docs = loader.load_and_split(text_splitter=text_splitter)

# 문서 분할 (2) : 이미 로드된 문서(docs)를 대상으로 텍스트 분할기를 적용
split_docs = text_splitter.split_documents(docs)

# 로드된 문서의 수 확인
print(len(split_docs))

# 다섯번째 문서 확인
split_docs[4]
```

## OpenAI 임베딩 생성 및 쿼리 결과 확인

```python
from langchain_openai import OpenAIEmbeddings

# OpenAI의 "text-embedding-3-small" 모델을 사용하여 임베딩을 생성합니다.
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")

# 텍스트를 임베딩하여 쿼리 결과를 생성합니다.
query_result = embeddings.embed_query("임베딩 테스트를 하기 위한 샘플 문장입니다.")
query_result[:5]
```

## 임베딩 대상 텍스트 정의

```python
# 임베딩 대상 텍스트
sentence1 = "안녕하세요? 반갑습니다."
sentence2 = "안녕하세요? 반갑습니다!"
sentence3 = "안녕하세요? 만나서 반가워요."
sentence4 = "Hi, nice to meet you."
sentence5 = "I like to eat apples."
```

## 임베딩 수행

```python
# 임베딩 수행
sentences = [sentence1, sentence2, sentence3, sentence4, sentence5]
embedded_sentences = embeddings.embed_documents(sentences)
```

## 코사인 유사도 계산 함수 정의

```python
from sklearn.metrics.pairwise import cosine_similarity

def similarity(a, b):
    return cosine_similarity([a], [b])[0][0]
```

## 유사도 계산 및 결과 출력

```python
# 유사도 계산
for i, sentence in enumerate(embedded_sentences):
    for j, other_sentence in enumerate(embedded_sentences):
        if i < j:
            print(
                f"[유사도 {similarity(sentence, other_sentence):.4f}] {sentences[i]} \t <=====> \t {sentences[j]}"
            )
```

## 질문-답변 프롬프트 템플릿 생성

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
    """당신은 질문-답변(Question-Answering)을 수행하는 친절한 AI 어시스턴트입니다. 당신의 임무는 주어진 문맥(context) 에서 주어진 질문(question) 에 답하는 것입니다.
검색된 다음 문맥(context) 을 사용하여 질문(question) 에 답하세요. 또한 어떤 문맥을 참고했느지 출처를 같이 표현해주세요.
만약, 주어진 문맥(context) 에서 답을 찾을 수 없다면, 답을 모른다면 주어진 정보에서 질문에 대한 정보를 찾을 수 없습니다 라고 답하세요.
한글로 답변해 주세요. 단, 기술적인 용어나 이름은 번역하지 않고 그대로 사용해 주세요.

#Question:
{question}

#Context:
{context}

#Answer:"""
)
```

## RAG 테스트

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model_name="gpt-5.4-mini")
retriever = db.as_retriever()
parser = StrOutputParser()
```

## LLM과 StrOutputParser 결합 사용

```python
print((llm | StrOutputParser()).invoke("온디바이스 AI 시장 규모에 대해 알려줘"))
```

## RAG 체인 사용하여 질문에 대한 응답 생성

```python
query = '온디바이스 AI 시장 규모에 대해 알려줘'
contexts = retriever.invoke(query)
chain = prompt | llm | StrOutputParser()
print(chain.invoke({'context': contexts, 'question': query}))
```

```python
# RunnablePassthrough를 이용한 방식
from langchain_core.runnables import RunnablePassthrough

rag_chain = (
    {"context": retriever, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)
print(rag_chain.invoke({'context': contexts, 'question': query}))
```

## 요청 재작성 (Query Re-write)

```python
from typing import TypedDict

class Schema(TypedDict):
    '''재작성된 query를 반환하기 위한 스키마'''
    rewrited_query: str

system_prompt = '''주어진 질문을 벡터 검색에 최적화된 형태로 재작성하세요.
# 재작성 지침:
1. 질문의 핵심 의도를 파악하세요
2. 명확하고 구체적인 키워드를 포함하세요
3. 검색에 도움이 되는 관련 용어를 추가하세요
4. 원래 질문과 같은 언어로 작성하세요'''

llm_rewrite = llm.with_structured_output(Schema) # 스키마 적용
query = "LLM 환각 문제 해결"
msgs = [('system', system_prompt), ('user', query)] # chat messages 형태로 전달
print(llm_rewrite.invoke(msgs)['rewrited_query']) # 재작성된 요청 확인
```

## 다중 요청 (Multi-query)

```python
class Schema(TypedDict):
    '''재작성된 queries를 반환하기 위한 스키마'''
    rewrited_queries: list[str]

system_prompt = '''다음 질문을 3가지 다른 관점에서 재작성하세요.
# 재작성 지침:
1. 질문의 핵심 의도를 파악하세요
2. 명확하고 구체적인 키워드를 포함하세요
3. 검색에 도움이 되는 관련 용어를 추가하세요
4. 원래 질문과 같은 언어로 작성하세요
5. 각 관점이 가급적 서로 겹치지 않도록 작성하세요.'''

llm_rewrite = llm.with_structured_output(Schema) # 스키마 적용
query = "LLM 환각 문제 해결"
msgs = [('system', system_prompt), ('user', query)] # chat messages 형태로 전달
print(llm_rewrite.invoke(msgs)['rewrited_queries']) # 재작성된 요청 확인
```

## HyDE

```python
class Schema(TypedDict):
    '''사용자 요청에 대한 생성된 가상 문서'''
    document: str

system_prompt = '''다음 질문에 대한 가상의 전문 문서를 작성하세요.
실제 문서처럼 구체적이고 상세하게 작성하세요.'''

llm_hyde = llm.with_structured_output(Schema) # 스키마 적용
query = "LLM 환각 문제 해결"
msgs = [('system', system_prompt), ('user', query)] # chat messages 형태로 전달
print(llm_hyde.invoke(msgs)['document'][:100]) # 가상 문서 확인
```

## 요청 분해 (Query Decomposition)

```python
class Schema(TypedDict):
    '''분해된 queries를 반환하기 위한 스키마'''
    decomposed_queries: list[str]

system_prompt = '''복잡한 질문을 2-4개의 간단한 하위 질문으로 분해하세요.
각 하위 질문은 독립적으로 검색 가능해야 합니다.'''

llm_decomposed = llm.with_structured_output(Schema) # 스키마 적용
query = "LLM 환각 문제의 원인, 해결 방법, 환각 관점에서의 LLM의 미래를 정리해줘"
msgs = [('system', system_prompt), ('user', query)] # chat messages 형태로 전달
print(llm_decomposed.invoke(msgs)['decomposed_queries']) # 분해된 요청 확인
```

## 셀프 요청 (Self-query) 검색기

```python
from langchain_classic.chains.query_constructor.base import AttributeInfo

metadata_field_info = [
    AttributeInfo(name="source", description="문서의 경로", type="string"),
    AttributeInfo(name="page_label", description="문서의 페이지 번호", type="string"),]

path1 = './data/온디바이스 AI 기술동향 및 발전방향.pdf'
path2 = './data/온디바이스 AI(On-Device AI) 산업현황 보고서.pdf'
db.similarity_search('온디바이스 AI 시장 동향', filter={'source': {'$eq': path1}})
```

```python
from langchain_classic.retrievers.self_query.base import SelfQueryRetriever
from langchain_community.query_constructors.chroma import ChromaTranslator

self_query_retriever = SelfQueryRetriever.from_llm(
    llm=llm, vectorstore=db,
    document_contents="온디바이스 AI 기술 동향 및 산업 현황",
    metadata_field_info=metadata_field_info,  # 메타데이터 스키마 전달
    structured_query_translator=ChromaTranslator(),
    verbose=True,  # 내부 쿼리 변환 과정 출력
    search_kwargs={'k':2})

queries = [f'{path1} 문서 내에서 온디바이스 AI 기술 동향을 검색해줘',
           f'{path2} 문서 내에서 온디바이스 AI 기술 동향을 검색해줘',]
for query in queries:
    results = self_query_retriever.invoke(query)
    print([result.metadata['source'] for result in results])
```

## 재순위화 (Re-ranking)

```python
from langchain_community.cross_encoders import HuggingFaceCrossEncoder
from langchain_classic.retrievers.document_compressors import CrossEncoderReranker
from langchain_classic.retrievers import ContextualCompressionRetriever

# 1차로 50개 추출
retriever = db.as_retriever(search_kwargs={'k': 50})

# 크로스 인코더 모델 생성
model = HuggingFaceCrossEncoder(model_name='BAAI/bge-reranker-v2-m3')
# 리랭커 생성
compressor = CrossEncoderReranker(model=model, top_n=5)

# base 검색기와 리랭커 결합
compression_retriever = ContextualCompressionRetriever(
    base_compressor=compressor,
    base_retriever=retriever)
```
