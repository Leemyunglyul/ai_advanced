## 실습 시작

```python
# rfp.pdf 파일의 로더를 생성하세요.
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("./data/rfp.pdf")
```

```python
# 문서를 로드한 후 로드된 문서(페이지)의 수를 출력하세요.
docs = loader.load()
len(docs)
```

```python
# 문서 내용을 출력하세요.
docs
```

```python
# 리스트 컴프리헨션
# 반복문을 이용한 리스트 생성
[doc.page_content for doc in docs]
```

```python
# 로드된 문서의 내용을 추출하여 문자열로 변환하세요.
# 페이지별 구분은 개행 두 개로 지정하세요.
context = '\n\n'.join([doc.page_content for doc in docs])
print(context)
```

```python
# 출력 스키마를 정의하세요.
# 출력 스키마는 자유롭게 정의해보세요.
from pydantic import BaseModel, Field
from typing import List, Optional

class RequirementNode(BaseModel):
    '''RFP의 개별 요구사항 항목 또는 섹션을 나타내는 노드'''
    title: str = Field(..., description='항목의 제목 (예: 시스템 성능 요구사항, 데이터 보안)')
    category: str = Field(..., description='항목의 유형 (예: 비즈니스 요구, 기술 요구, 관리 요구, 일반 사항)')
    content: str = Field(..., description='해당 항목의 구체적인 본문 내용 또는 요약')
    importance: str = Field(..., description='중요도 또는 우선순위 (필수, 선택, 해당없음 등)')

    sub_requirements: List['RequirementNode'] = Field(default_factory=list,
                                                      description='해당 항목에 속한 하위 세부 요구사항들')

class RFPAnalysisResult(BaseModel):
    '''RFP 분석 결과의 최상위 루트 객체'''
    project_name: str = Field(..., description='추출된 프로젝트 명칭')
    overview: str = Field(..., description='전체 사업 개요 요약')
    total_budget: Optional[str] = Field(None, description='명시된 경우 총 예산 규모')
    deadline: Optional[str] = Field(None, description='사업 기간 또는 제안서 제출 마감일')

    requirements_tree: List[RequirementNode] = Field(..., description='RFP에서 추출된 계층 구조의 요구사항 리스트')
```

```python
# LLM을 생성하세요.
# gpt-5.4-mini 모델을 사용하세요.
# 모델의 동작을 확인하세요.
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model='gpt-5.4-mini')
llm.invoke('안녕')
```

```python
# LLM에 출력 스키마를 바인딩하세요.
llm_with_schema = llm.with_structured_output(RFPAnalysisResult)
llm_with_schema
```

```python
# 프롬프트 템플릿 생성하세요.
# rfp 입력을 받을 수 있도록 자유롭게 생성해보세요.
from langchain_core.prompts import PromptTemplate

template = '''너는 좋은 AI 도우미야.
아래 전달된 RFP 문서(RFP Document)를 세부적으로 상세히 분석해서 바인딩된 출력 스키마에 맞게 추출해줘.
요구사항 내에 포함된 세부 항목(불렛 포인트 등)이 있다면 반드시 sub_requirements에 계층적으로 담아줘.
모든 내용은 한국어로 번역해서 생성해줘.

# RFP Document:
{document}
'''

prompt = PromptTemplate.from_template(template)
prompt
```

```python
# 체인을 구성한 후 출력을 생성하세요.
chain = prompt | llm_with_schema
result = chain.invoke({
    'document': context
})
```
